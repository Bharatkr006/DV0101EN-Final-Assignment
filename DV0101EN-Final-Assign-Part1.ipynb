## Historical Automobile Sales Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import folium
%matplotlib inline

df = pd.read_csv('historical_automobile_sales.csv')
print('Data loaded successfully!')
```

### TASK 1.1: Develop a Line chart showing how automobile sales fluctuate from year to year

```python
df_line = df.groupby('Year')['Automobile_Sales'].mean().reset_index()
plt.figure(figsize=(10, 6))
plt.plot(df_line['Year'], df_line['Automobile_Sales'], marker='o')
plt.title('Automobile Sales Fluctuation Year to Year')
plt.xlabel('Year')
plt.ylabel('Automobile Sales')
plt.xticks(list(range(1980, 2024, 2)), rotation=45)
plt.grid(True)
plt.show()
```

### TASK 1.2: Plot different lines for categories of vehicle type during recession and non-recession periods

```python
df_mline = df.groupby(['Year', 'Vehicle_Type'])['Automobile_Sales'].mean().reset_index()
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_mline, x='Year', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Automobile Sales Trends per Vehicle Type')
plt.xlabel('Year')
plt.ylabel('Automobile Sales')
plt.grid(True)
plt.show()
```

### TASK 1.3: Use Seaborn to visualize sales trends per vehicle type during recession vs non-recession

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Vehicle_Type', y='Automobile_Sales', hue='Recession', data=df)
plt.title('Vehicle Sales Comparison During Recession vs Non-Recession')
plt.xlabel('Vehicle Type')
plt.ylabel('Automobile Sales')
plt.show()
```

### TASK 1.4: Subplots for GDP variations during recession vs non-recession

```python
rec_data = df[df['Recession'] == 1]
non_rec_data = df[df['Recession'] == 0]
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))
sns.lineplot(x='Year', y='GDP', data=rec_data, ax=ax1, label='Recession GDP')
ax1.set_title('GDP Variation During Recession')
sns.lineplot(x='Year', y='GDP', data=non_rec_data, ax=ax2, label='Non-Recession GDP')
ax2.set_title('GDP Variation During Non-Recession')
plt.tight_layout()
plt.show()
```

### TASK 1.5: Bubble plot for seasonality impact on Automobile Sales

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Month', y='Automobile_Sales', size='Seasonality_Weight', hue='Seasonality_Weight', sizes=(20, 200))
plt.title('Seasonality Impact on Automobile Sales')
plt.xlabel('Month')
plt.ylabel('Automobile Sales')
plt.show()
```

### TASK 1.6: Scatter plot showing correlation between average vehicle price and sales volume during recessions

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=rec_data, x='Price', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Vehicle Price vs Automobile Sales Volume During Recession')
plt.xlabel('Average Vehicle Price')
plt.ylabel('Automobile Sales Volume')
plt.show()
```

### TASK 1.7: Pie chart showing advertisement expenditure share during recession vs non-recession

```python
R_exp = rec_data['Advertising_Expenditure'].sum()
NR_exp = non_rec_data['Advertising_Expenditure'].sum()
plt.figure(figsize=(8, 8))
plt.pie([R_exp, NR_exp], labels=['Recession', 'Non-Recession'], autopct='%1.1f%%', startangle=90)
plt.title('Total Advertisement Expenditure Share')
plt.show()
```

### TASK 1.8: Pie chart showing total advertisement expenditure spent per vehicle type during recession

```python
VT_exp = rec_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()
plt.figure(figsize=(8, 8))
plt.pie(VT_exp, labels=VT_exp.index, autopct='%1.1f%%', startangle=90)
plt.title('Advertisement Expenditure per Vehicle Type During Recession')
plt.show()
```

### TASK 1.9: Line plot showing effect of unemployment rate on vehicle type and sales

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=rec_data, x='unemployment_rate', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Effect of Unemployment Rate on Vehicle Sales During Recession')
plt.xlabel('Unemployment Rate')
plt.ylabel('Automobile Sales')
plt.show()
```
